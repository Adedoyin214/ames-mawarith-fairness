## Fairness Metrics in Islamic Inheritance Distribution
**Measures:** Gini Coefficient, Theil T Index, Atkinson Index, Coefficient of Variation  
**Dataset:** 20 PCIED cases + 5,000 simulation records  
**Benchmark:** Civil equal-share distribution  
**Institute:** Elbaseety Institute® | Mawārith Analytics  
**Source Data:** PCIED 2023 — Alhikma University, Ilorin


```python
import json, statistics as st, collections
import matplotlib.pyplot as plt
import numpy as np
import sys; sys.path.insert(0, '.')
from fairness_analysis import gini, theil_t, atkinson, coefficient_of_variation

pcied = json.load(open('pcied_fairness.json'))
sim   = json.load(open('sim_fairness.json'))
print(f"Loaded: {len(pcied)} PCIED cases, {len(sim):,} simulation records")
```

## 1. PCIED Case Fairness Metrics

```python
print(f"{'ID':>3}  {'Gini':>8}  {'Theil T':>8}  {'Atkinson':>9}  {'CV':>7}  Type")
print("-"*55)
for p in pcied:
    t = "Tanzeel" if p['tanzeel'] else "'Awl" if p['awl'] else "Radd" if p['radd'] else "Std"
    print(f"{p['case_id']:>3}  {p['islamic']['gini']:>8.4f}  {p['islamic']['theil_t']:>8.4f}  "
          f"{p['islamic']['atkinson']:>9.4f}  {p['islamic']['cv']:>7.4f}  {t}")
print()
ginis=[p['islamic']['gini'] for p in pcied]
print(f"Mean Gini (Islamic) : {st.mean(ginis):.4f}")
print(f"Mean Gini (Civil)   : {st.mean(p['civil']['gini'] for p in pcied):.4f}")
```

## 2. Simulation Summary

```python
ginis_isl=[r['gini_islamic'] for r in sim]
theils=[r['theil_islamic'] for r in sim]
atkinsons=[r['atkinson_islamic'] for r in sim]
print(f"Gini  — Mean: {st.mean(ginis_isl):.4f}  Median: {st.median(ginis_isl):.4f}  Std: {st.stdev(ginis_isl):.4f}")
print(f"Theil — Mean: {st.mean(theils):.4f}  Median: {st.median(theils):.4f}")
print(f"Atkinson Mean: {st.mean(atkinsons):.4f}")
print()
by_profile=collections.defaultdict(list)
for r in sim: by_profile[r['profile']].append(r['gini_islamic'])
print("Mean Gini by Profile:")
for p,v in sorted(by_profile.items(), key=lambda x: -st.mean(x[1])):
    print(f"  {p:<45} {st.mean(v):.4f}")
```

## 4. Key Findings

| Metric | Islamic (sim) | Islamic (PCIED) | Civil Benchmark |
|--------|:------------:|:---------------:|:---------------:|
| Gini | 0.1949 | 0.2281 | 0.0000 |
| Theil T | 0.0760 | 0.1310 | 0.0000 |
| Atkinson | 0.0390 | 0.0680 | 0.0000 |

- Islamic inheritance is **moderately unequal by design** — differential shares between heirs reflect their varying obligations and relationships.
- Civil equal-share achieves Gini = 0 mathematically, but ignores role-based differentiation.
- Complex family structures produce the highest inequality (Gini ≈ 0.258).
- Case 7 (Gini = 0.000) is the only perfectly equal PCIED case — a rare outcome after Radd.
